# Streaming updates

This notebook sends a surface in pieces. It first sends the component tree with an empty list, then appends restaurants one at a time, then changes an existing item and adds a footer. Each `send` call is validated and applied in order, so the page fills in while the cell runs.

```python
import sys
from importlib.util import find_spec

# Pyodide and xeus-python both report sys.platform "emscripten"; only Pyodide ships the pyodide package.
if sys.platform != "emscripten":
    KERNEL = "cpython"
else:
    try:
        has_pyodide = find_spec("pyodide") is not None
    except ValueError:  # xeus-python's pyodide stub has __spec__ set to None
        has_pyodide = False
    KERNEL = "pyodide" if has_pyodide else "xpython"

if KERNEL == "pyodide" and find_spec("a2ui_jupyter") is None:
    import piplite

    await piplite.install(["rpds-py", "a2ui-jupyter"])

from a2ui_jupyter import A2UISurface

BASIC_CATALOG_ID = "https://a2ui.org/specification/v1_0/catalogs/basic/catalog.json"


def new_surface(surface_id, *, send_data_model=False, **widget_kwargs):
    surface = A2UISurface(**widget_kwargs)
    create = {"surfaceId": surface_id, "catalogId": BASIC_CATALOG_ID}
    if send_data_model:
        create["sendDataModel"] = True
    surface.send({"version": "1.0", "createSurface": create})
    return surface
```

```python
surface = new_surface("streaming-updates")
surface.send(
    [
        {
            "version": "1.0",
            "updateDataModel": {
                "surfaceId": "streaming-updates",
                "value": {"restaurants": []},
            },
        },
        {
            "version": "1.0",
            "updateComponents": {
                "surfaceId": "streaming-updates",
                "components": [
                    {
                        "id": "root",
                        "component": "Column",
                        "children": ["heading", "restaurant_list"],
                    },
                    {"id": "heading", "component": "Text", "text": "## Restaurants"},
                    {
                        "id": "restaurant_list",
                        "component": "Column",
                        "children": {"path": "/restaurants", "componentId": "restaurant_card"},
                    },
                    {
                        "id": "restaurant_card",
                        "component": "Column",
                        "children": ["card_title", "card_subtitle"],
                    },
                    {"id": "card_title", "component": "Text", "text": {"@path": "title"}},
                    {
                        "id": "card_subtitle",
                        "component": "Text",
                        "text": {"@path": "subtitle"},
                        "variant": "caption",
                    },
                ],
            },
        },
    ]
)
surface
```

This cell appends one restaurant every second. Each `updateDataModel` targets the next index in `/restaurants`, and the list template adds a card for it.

```python
import asyncio

restaurants = [
    {"title": "The Golden Fork", "subtitle": "Fine dining and spirits"},
    {"title": "Ocean's Bounty", "subtitle": "Fresh daily seafood"},
    {"title": "Pizzeria Roma", "subtitle": "Wood-fired pizza"},
]

for index, restaurant in enumerate(restaurants):
    surface.send(
        {
            "version": "1.0",
            "updateDataModel": {
                "surfaceId": "streaming-updates",
                "path": f"/restaurants/{index}",
                "value": restaurant,
            },
        }
    )
    await asyncio.sleep(1)
```

This cell changes the subtitle of the first restaurant in place and adds a footer. Sending `updateComponents` again with the same ids replaces those components and the root's child list.

```python
surface.send(
    [
        {
            "version": "1.0",
            "updateDataModel": {
                "surfaceId": "streaming-updates",
                "path": "/restaurants/0/subtitle",
                "value": "Fine dining, now open late",
            },
        },
        {
            "version": "1.0",
            "updateComponents": {
                "surfaceId": "streaming-updates",
                "components": [
                    {
                        "id": "root",
                        "component": "Column",
                        "children": ["heading", "restaurant_list", "footer"],
                    },
                    {
                        "id": "footer",
                        "component": "Text",
                        "text": "Updated in place by a later cell",
                        "variant": "caption",
                    },
                ],
            },
        },
    ]
)
```
